```python
import os
import pickle
import networkx as nx
import matplotlib.pyplot as plt

def visualize_and_save_graphs(input_folder, output_folder):
    """
    Reads .pkl graph files from the input_folder, generates a plot for each,
    and saves the images to the output_folder.
    """
    # Create the output directory if it doesn't exist
    os.makedirs(output_folder, exist_ok=True)
    
    # Check if input directory exists
    if not os.path.exists(input_folder):
        print(f"Folder '{input_folder}' does not exist. Skipping.")
        return
        
    print(f"Processing graphs from '{input_folder}' to '{output_folder}'...")
    
    # Process each pickle file
    for filename in sorted(os.listdir(input_folder)):
        if filename.endswith('.pkl'):
            file_path = os.path.join(input_folder, filename)
            
            # Load the graph data
            with open(file_path, 'rb') as f:
                data = pickle.load(f)
                
            # Reconstruct the Graph
            G = nx.Graph()
            # Adding nodes first ensures isolated vertices are included
            G.add_nodes_from(range(data['num_vertices']))
            G.add_edges_from(data['edge_list'])
            
            # Create a new figure (important for memory management inside a loop)
            fig = plt.figure(figsize=(8, 6))
            
            # Use a spring layout for general graph visualization
            pos = nx.spring_layout(G, seed=42)
            
            # Add a descriptive title
            title = f"ID: {data['id']} | Type: {data['type']} | Nodes: {data['num_vertices']} | Edges: {data['num_edges']}"
            plt.title(title, fontsize=12, fontweight='bold')
            
            # Draw the graph
            nx.draw(G, pos, 
                    with_labels=True, 
                    node_color='skyblue', 
                    edge_color='gray', 
                    node_size=400, 
                    font_size=8, 
                    font_weight='bold')
            
            # Construct the output image filename
            image_filename = filename.replace('.pkl', '.png')
            image_path = os.path.join(output_folder, image_filename)
            
            # Save the figure
            plt.savefig(image_path, format='png', bbox_inches='tight')
            
            # Close the figure to free up memory (crucial when generating thousands of images)
            plt.close(fig)
            
    print(f"Finished saving images to '{output_folder}'.\n")

# Execute the visualization for both datasets
visualize_and_save_graphs('train', 'train_image_barabasi_albert')
visualize_and_save_graphs('test', 'test_image_barabasi_albert')
```
